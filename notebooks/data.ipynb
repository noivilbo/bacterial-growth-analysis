```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from src.data_cleaning import load_data,save_data,clean_data
df = load_data("data/raw/bacteria_growth.csv")
cleaned_data = clean_data(df)
save_data(cleaned_data,"data/processed/cleaned_data.csv")
```

```python
cleaned_data.head()
cleaned_data.info()
cleaned_data.describe()
```

```python
sns.lineplot(
    data = cleaned_data,
    x = "temperature",
    y = "capacity",
    hue = "strain"
)
```

### Temperature
Each strain shows a different optimal temperature.
Strains A performs better at around 25c,strain B at 30c and C near 37c.This suggests that growth is directly correlated to ambient temperature.

```python
sns.lineplot(
    data = cleaned_data,
    x = "hour",
    y = "population",
    hue = "strain"
)
```

### Observation
Strain B as the fastest growth and reaches the highest population.
Strain C grows slower and stabilize in a smaller population
